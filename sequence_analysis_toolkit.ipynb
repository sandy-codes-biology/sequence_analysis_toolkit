# Sequence Analysis Toolkit
This notebook takes a DNA sequence and reports:
- 📏 Sequence length
- 🧬 GC content
- 🔄 Reverse complement
- 🟢 Translation to protein sequence
- 📖 ORF finding (forward strand, frame 1 only)

```python
def sequence_length(seq):
    return len(seq)

seq= input("Enter the sequence:").strip().upper() # .strip() and .upper() to avoid errors when inputting DNA sequences
print("Length:", sequence_length(seq))
```

```python
def gc_content(seq):
    g = seq.count("G")
    c = seq.count("C")
    return (g+c) / len(seq) * 100

print("GC content as a %:", gc_content(seq))
```

```python
def reverse_complement(seq):
    complement = {"A":"T","T":"A","G":"C","C":"G"}
    comp = ""
    for base in seq:
        comp = comp + complement[base]
    return comp[::-1]

print("The reverse complement of this sequence is:", reverse_complement(seq))
```

```python
codon_table = {
    "TTT": "F", "TTC": "F", "TTA": "L", "TTG": "L",
    "CTT": "L", "CTC": "L", "CTA": "L", "CTG": "L",
    "ATT": "I", "ATC": "I", "ATA": "I", "ATG": "M",
    "GTT": "V", "GTC": "V", "GTA": "V", "GTG": "V",
    "TCT": "S", "TCC": "S", "TCA": "S", "TCG": "S",
    "CCT": "P", "CCC": "P", "CCA": "P", "CCG": "P",
    "ACT": "T", "ACC": "T", "ACA": "T", "ACG": "T",
    "GCT": "A", "GCC": "A", "GCA": "A", "GCG": "A",
    "TAT": "Y", "TAC": "Y", "TAA": "*", "TAG": "*",
    "CAT": "H", "CAC": "H", "CAA": "Q", "CAG": "Q",
    "AAT": "N", "AAC": "N", "AAA": "K", "AAG": "K",
    "GAT": "D", "GAC": "D", "GAA": "E", "GAG": "E",
    "TGT": "C", "TGC": "C", "TGA": "*", "TGG": "W",
    "CGT": "R", "CGC": "R", "CGA": "R", "CGG": "R",
    "AGT": "S", "AGC": "S", "AGA": "R", "AGG": "R",
    "GGT": "G", "GGC": "G", "GGA": "G", "GGG": "G"
}

def translate(seq):
    protein=""

    for i in range(0, len(seq)-2, 3): # range tells python to visit the sequence at specific positions, e.g. 0, 3, 6
        codon = seq[i:i+3] # extracts 3 letters at every 'i' position
        amino_acid = codon_table[codon] # go to dictionary 'codon_table' and find key value associated with the 3 letters
        protein= protein + amino_acid
    return protein

print("The codon sequence is:",translate(seq))
```

```python
def find_orf(seq):
    orfs = []
    stop_codons = ["TAA", "TAG", "TGA"]

    # Search for START codons (ATG)
    for i in range(0, len(seq) - 2, 3):
        codon = seq[i:i+3]

        # Look for a stop codon after the START codon
        if codon == "ATG":
            for j in range(i + 3, len(seq) - 2, 3):
                stop_codon = seq[j:j+3]

                if stop_codon in stop_codons:
                    orf = seq[i:j+3] # +3 so it includes the whole STOP codon
                    orfs.append(orf)
                    break

    return orfs

print("The ORF is:", find_orf(seq))
```
